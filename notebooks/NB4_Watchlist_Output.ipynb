```python
# -------------------------------------------------------
# Step 2: Add Placeholder Importance Scores
# -------------------------------------------------------
# Person 3 has not finished yet so we use random scores
# between 0 and 1 as placeholders.
# Replace this entire cell when Person 3 provides their output!

np.random.seed(42)  # makes results reproducible
df['importance_score'] = np.random.uniform(0, 1, len(df)).round(2)

# Convert to label
def importance_label(score):
    if score >= 0.67:
        return 'High'
    elif score >= 0.33:
        return 'Medium'
    else:
        return 'Low'

df['importance_label'] = df['importance_score'].apply(importance_label)

print(df['importance_label'].value_counts())
```
